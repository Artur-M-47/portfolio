## 🛠️ Methodology and Pipeline
1. **Data Preprocessing and Feature Engineering:** Extensive cleaning, including handling rare categories, imputing missing values, **normalizing** numerical features, and using **One-Hot Encoding (OHE)** to transform the feature space into **107 dimensions.**
2. **Tiered Modeling Strategy:** Implementation and comparison of three distinct machine learning approaches:

    - **Baseline: Logistic Regression** (Applied to Scaled OHE data).
    - **Ensemble: CatBoost Classifier** (Baseline and Tuned versions).
    - **Deep Learning: Deep Neural Network (DNN)** with Keras (Applied to Scaled OHE data).  

3. **Hyperparameter Tuning:** Focused tuning (especially for CatBoost) using a dual-criteria objective to select the best model: maximizing MCC and using minimizing Logloss as a tie-breaker.
4. **Evaluation:** Final assessment using key metrics robust for classification: **Matthews Correlation Coefficient (MCC)** and **Log Loss**.
